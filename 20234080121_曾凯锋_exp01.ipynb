## 二、实验1：基本运算、数据类型与字符串操作


### 1. 算术运算符与结果类型


````python
print("8 / 3 =", 8 / 3, type(8 / 3))    # /  真除法，结果始终为 float
print("8 // 3 =", 8 // 3, type(8 // 3)) # // 整除（向下取整除法），结果为 int
print("8 % 3 =", 8 % 3, type(8 % 3))    # %  取余数，结果为 int
print("2 ** 7 =", 2 ** 7, type(2 ** 7)) # ** 幂运算，结果为 int
````

**说明**：`/` 是真除法，即使整除结果也是 `float`；`//` 是整除；`%` 取余；`**` 是幂运算。


### 2. 创建四种基本类型并用 type() 检查


````python
int_val = 42
float_val = 3.14
str_val = "hello"
bool_val = True
print(type(int_val), type(float_val), type(str_val), type(bool_val))
````

### 3. 类型转换与异常捕获


````python
print(int("123"))     # 字符串 -> 整数：123
print(float("3.14"))  # 字符串 -> 浮点数：3.14
print(int(3.9))       # 浮点数 -> 整数：向零截断，得 3（不是四舍五入）

try:
    int("goat")
except ValueError as e:
    print("捕获异常:", type(e).__name__, "-", e)
````

### 4. 字符串处理：去空白、大小写、切片、拆分与 join

原始字符串：`" Python Data Analysis "`（首尾各有一个空格）。


````python
raw = " Python Data Analysis "
text = raw.strip()            # 去除首尾空白
print(repr(text))
print(text.upper())           # 全大写
print(text.lower())           # 全小写
print(text[:6], "|", text[-8:])  # 切片：前 6 个字符 与 后 8 个字符
words = text.split()          # 默认按空白拆分单词
print(words)
print(" ".join(words))        # 用空格连接
print("-".join(words))        # 用连字符连接
````

````python
print("goat" + "max")   # 拼接
print("goat" * 3)       # 重复

try:
    text[0] = "J"       # 字符串不可变，直接修改单个字符会报错
except TypeError as e:
    print("捕获异常:", type(e).__name__, "-", e)

print("J" + text[1:])   # 正确方式：基于切片构造新字符串
````

## 三、实验2：函数、条件、循环与递归


### 1. score_level(score)：分段评级 + ValueError


````python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"
````

````python
for sc in (95, 85, 76, 59):
    print(sc, score_level(sc))

for sc in (105, -1):  # 超出范围的数值抛出 ValueError
    try:
        score_level(sc)
    except ValueError as e:
        print(f"{sc} 捕获异常: {e}")
````

### 2. 分别用 for 和 while 计算 1 到 100 的整数和


````python
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print(total_for, total_while, total_for == total_while)  # 两种实现结果一致：5050
````

### 3. 递归版与迭代版阶乘（约定 0! = 1）


````python
def factorial_rec(n):
    if n == 0:
        return 1
    return n * factorial_rec(n - 1)


def factorial_iter(n):
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result
````

````python
for n in (0, 1, 5, 10):
    print(n, factorial_rec(n), factorial_iter(n), factorial_rec(n) == factorial_iter(n))
````

### 4. summary(values)：以元组返回最小值、最大值、平均值


````python
def summary(values):
    return min(values), max(values), sum(values) / len(values)


lo, hi, avg = summary([85, 58, 92, 76, 60, 89])  # 元组解包接收结果
print(lo, hi, avg)
````

### 5. sum_all(*args)：可变参数求和


````python
def sum_all(*args):
    total = 0
    for x in args:
        total += x
    return total


print(sum_all())           # 无参数：0
print(sum_all(5))          # 单个参数：5
print(sum_all(1, 2, 3, 4)) # 多个参数：10
````

````python
demo = [85, 58, 92]
alias = demo          # 别名：与 demo 指向同一个列表对象
copied = demo.copy()  # 副本：创建一个新列表对象
alias[0] = 100        # 通过别名修改
print(demo)                     # 原列表被改变
print(alias)                    # 别名当然同步改变
print(copied)                   # 副本不受影响
print(alias is demo, copied is demo)  # True False：别名同一对象，副本是新对象
````

### 2. 写入 students.txt，再读取并跳过表头


````python
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
file_path = os.path.join(data_dir, "students.txt")
with open(file_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(file_path, encoding="utf-8") as f:
    lines = f.read().splitlines()
for line in lines[1:]:  # 跳过表头
    print(line)
````

### 3. parse_student(line)：解析单行记录，无效记录抛出 ValueError

检查三件事：字段数量必须为 2、姓名不能为空、成绩必须是 0~100 的整数。


````python
def parse_student(line):
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误：期望 2 个，实际 {len(parts)} 个")
    name, score_str = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError("姓名为空")
    score = int(score_str)  # 非数字字符串会自然抛出 ValueError
    if not 0 <= score <= 100:
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
    return {"name": name, "score": score}
````

### 4. 按行处理：捕获异常、记录行号与原因、收集有效记录


````python
valid = []    # 有效记录（字典列表）
errors = []   # (行号, 错误原因)
for lineno, line in enumerate(lines[1:], start=2):  # 数据从第 2 行开始
    line = line.strip()
    if not line:
        continue
    try:
        valid.append(parse_student(line))
    except ValueError as e:
        errors.append((lineno, str(e)))
        continue  # 继续处理后续记录
print(valid)
print(errors)
````

### 5. 写 report.txt（w 与 a 模式对比）

- `w` 模式：打开即清空，从头写入（覆盖已有内容）
- `a` 模式：追加，在文件末尾写入，已有内容保留


````python
n_valid = len(valid)
n_invalid = len(errors)
avg = sum(r["score"] for r in valid) / n_valid
pass_rate = sum(1 for r in valid if r["score"] >= 60) / n_valid

report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:  # w 模式：覆盖写入
    f.write(f"有效记录数：{n_valid}\n无效记录数：{n_invalid}\n平均分：{avg:.2f}\n及格率：{pass_rate:.2%}\n")
print(open(report_path, encoding="utf-8").read())

with open(report_path, "a", encoding="utf-8") as f:  # a 模式：追加一行
    f.write("（本行由 a 模式追加，原有内容保留）\n")
print(open(report_path, encoding="utf-8").read())
````

## 六、实验5：类、继承与对象复制


### 1. 定义 Student 类


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


stu = Student("Alice", 85)
print(stu, stu.is_passed())
stu2 = Student("Bob", 58)
print(stu2, stu2.is_passed())
````

### 2. 为实验4的每条有效记录创建 Student 对象


````python
students = [Student(r["name"], r["score"]) for r in valid]
for st in students:
    print(st, "及格" if st.is_passed() else "不及格")
````

### 3. GraduateStudent 继承 Student，重写 __str__


````python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)  # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"topic={self.research_topic})")


g = GraduateStudent("Eva", 88, "深度学习")
print(g, g.is_passed())  # is_passed 继承自 Student
````

### 4. 直接赋值 vs copy.copy vs copy.deepcopy

构造包含嵌套列表的对象，每次比较都从新建的原始对象开始。


````python
import copy

# 直接赋值：同一对象的两个引用，修改内部列表会影响原对象
original = {"name": "Alice", "scores": [85, 90]}
d1 = original
d1["scores"][0] = 0
print("直接赋值:", original)

# 浅拷贝：外层是新对象，但内部列表仍是共享引用
original = {"name": "Alice", "scores": [85, 90]}
d2 = copy.copy(original)
d2["scores"][0] = 0
print("浅拷贝:  ", original)

# 深拷贝：外层与内层都彻底复制，互不影响
original = {"name": "Alice", "scores": [85, 90]}
d3 = copy.deepcopy(original)
d3["scores"][0] = 0
print("深拷贝:  ", original)
````

**结论**：直接赋值不产生新对象；浅拷贝只复制最外层，嵌套的可变对象仍共享；
深拷贝递归复制所有层级，修改副本完全不影响原对象。
